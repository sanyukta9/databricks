```sql
WITH country_stats AS (
  SELECT
    country,
    COUNT(*) AS circuit_count,
    AVG(alt) AS avg_altitude
  FROM
    default.circuits
  GROUP BY
    country
),
top_countries AS (
  SELECT
    country
  FROM
    country_stats
  ORDER BY
    circuit_count DESC
  LIMIT 5
)
SELECT
  c.country,
  c.circuitId,
  c.name,
  c.location,
  c.lat,
  c.lng,
  c.alt,
  cs.circuit_count,
  cs.avg_altitude
FROM
  default.circuits c
JOIN
  country_stats cs ON c.country = cs.country
JOIN
  top_countries tc ON c.country = tc.country
ORDER BY
  cs.circuit_count DESC, c.country, c.name
```
